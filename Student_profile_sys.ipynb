```python
print("===== ENTER STUDENT INFORMATION =====")

raw_name    = input("Full Name:                 ")   
raw_age     = input("Age:                       ")
raw_gpa     = input("GPA:                       ")   
raw_courses = input("Courses (comma-separated): ")   


name    : str   = raw_name.strip()                   
age     : int   = int(raw_age)                       
gpa     : float = float(raw_gpa)                     
courses : list  = [c.strip() for c in raw_courses.split(",")]  

print(f"  raw_age  : {raw_age!r}  →  int({raw_age}) = {age}   (type: {type(age).__name__})")
print(f"  raw_gpa  : {raw_gpa!r}  →  float({raw_gpa}) = {gpa} (type: {type(gpa).__name__})")
print(f"  raw_courses : {raw_courses!r}")
print(f"            →  {courses}  (type: {type(courses).__name__})")


is_deans_list : bool = gpa >= 3.5      
is_minor      : bool = age < 18        


student_profile : dict = {
    "name"       : name,
    "age"        : age,
    "gpa"        : gpa,
    "courses"    : courses,
    "num_courses": len(courses),
}


age_status   = "Minor"  if is_minor      else "Adult"
deans_status = "Yes"    if is_deans_list else "No"

print("\n===== STUDENT PROFILE =====")
print(f"Name:        {student_profile['name']}")
print(f"Age:         {student_profile['age']} ({age_status})")
print(f"GPA:         {student_profile['gpa']:.2f}")
print(f"Courses:     {student_profile['courses']}")
print(f"# Courses:   {student_profile['num_courses']}")
print(f"Dean's List: {deans_status}")
print("===========================")


print(f"  name        → {type(student_profile['name']).__name__}")
print(f"  age         → {type(student_profile['age']).__name__}")
print(f"  gpa         → {type(student_profile['gpa']).__name__}")
print(f"  courses     → {type(student_profile['courses']).__name__}")
print(f"  is_deans_list → {type(is_deans_list).__name__}  (value: {is_deans_list})")
print(f"  student_profile → {type(student_profile).__name__}")
```
