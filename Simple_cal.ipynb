```python
print("===== SIMPLE CALCULATOR =====")

num1     = float(input("Enter first number:                    "))
operator = input("Enter operator (+, -, *, /, //, %, **): ").strip()
num2     = float(input("Enter second number:                   "))

result = None
error  = None

if operator == "+":
    result = num1 + num2

elif operator == "-":
    result = num1 - num2

elif operator == "*":
    result = num1 * num2

elif operator == "/":
    if num2 == 0:
        error = "Error: Cannot divide by zero."
    else:
        result = num1 / num2

elif operator == "//":
    if num2 == 0:
        error = "Error: Cannot floor-divide by zero."
    else:
        result = num1 // num2

elif operator == "%":
    if num2 == 0:
        error = "Error: Cannot use modulo with zero."
    else:
        result = num1 % num2

elif operator == "**":
    result = num1 ** num2

else:
    error = f"Error: '{operator}' is not a valid operator."

print()
if error:
    print(error)
else:
    
    if operator in ("/", "//", "%"):
        print(f"Result: {num1} {operator} {num2} = {result:.4f}")
    elif operator == "**":
        print(f"Result: {num1} {operator} {num2} = {result:g}")
    else:
        print(f"Result: {num1} {operator} {num2} = {result:g}")

    if result > 0:
        print("The result is positive.")
    elif result < 0:
        print("The result is negative.")
    else:
        print("The result is zero.")

print("==============================")
```
